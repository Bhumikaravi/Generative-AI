```python
#Import the required library
from transformers import pipeline
```

```python
#Load GPT-2
generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
#Create the prompts
prompts = [
    "A doctor is",
    "A nurse is",
    "A business executive is",
    "A scientist is",
    "A police officer is",
    "A criminal is"
]

print(prompts)
```

```python
#Generate 5 outputs for each profession
results = {}

for prompt in prompts:
    print("\n" + "=" * 60)
    print("PROMPT:", prompt)

    outputs = generator(
        prompt,
        max_new_tokens=40,
        num_return_sequences=5,
        do_sample=True,
        temperature=0.8
    )

    results[prompt] = outputs

    for i, output in enumerate(outputs, 1):
        print(f"{i}. {output['generated_text']}")
```
